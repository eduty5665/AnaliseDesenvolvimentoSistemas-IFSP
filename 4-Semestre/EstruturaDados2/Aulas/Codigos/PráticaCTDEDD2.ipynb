```python
import time
import matplotlib.pyplot as plt

# Funções com diferentes complexidades
def constante(n):
    return n + 1  # operação constante, independe do tamanho de n

def linear(n):
    soma = 0
    for i in range(n):
        soma += i
    return soma

def quadratica(n):
    soma = 0
    for i in range(n):
        for j in range(n):
            soma += i * j
    return soma



# Medir tempo de execução
def medir_tempo(func, entradas):
    tempos = []
    for n in entradas:
        inicio = time.time()
        func(n)
        fim = time.time()
        tempos.append(fim - inicio)
    return tempos

# Lista de entradas
entradas = [10, 100, 500, 1000, 2000, 3000,5000]


# Medir os tempos de cada função
tempos_constante = medir_tempo(constante, entradas)
tempos_linear = medir_tempo(linear, entradas)
tempos_quadratica = medir_tempo(quadratica, entradas)


plt.figure(figsize=(10, 6))
plt.plot(entradas, tempos_constante, 'ro-', label='O(1) - Constante')
plt.plot(entradas, tempos_linear, 'go-', label='O(n) - Linear')
plt.plot(entradas, tempos_quadratica, 'bo-', label='O(n²) - Quadrática')

plt.xlabel('Tamanho da entrada (n)')
plt.ylabel('Tempo de execução (s)')
plt.title('Tempo de execução vs Tamanho da entrada')
plt.legend()
#plt.yscale('log')
plt.grid(True)
plt.show()
```
